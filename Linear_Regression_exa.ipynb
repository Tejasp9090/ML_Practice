```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Salary_Data.csv')
df[:2]
```

```python
plt.scatter(df['YearsExperience'],df['Salary'])
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
# Seprate data
x = df[['YearsExperience']]
y = df['Salary']
```

```python
x.ndim,y.ndim
```

```python
# Import algorithm
from sklearn.linear_model import LinearRegression
```

```python
# Create a model (Object)
model = LinearRegression()
model
```

```python
# Trainning -fit(input,output)
model.fit(x,y)
```

```python
# Testing
model.predict([[2.7]])
```

```python
test = [[0],[2.5],[5.7],[7],[12]]

pred = model.predict(test)
pred
```

```python
plt.scatter(x,y)
plt.plot(test,pred,color='red',label='Prediction')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```
